You can adjust the period parameter to get different time ranges:
* '1d': 1 day
* '5d': 5 days
* '1mo': 1 month
* '1y': 1 year
* 'max': maximum available data.

```python
import yfinance as yf

# Define the ticker symbol
ticker_symbol = ["AAPL","MSFT","NVDA","JPM","XOM","JNJ","KO","WMT"]

# Download historical data 
data = yf.download(ticker_symbol, period="10y", group_by="ticker", auto_adjust=True)

# Download the tickers as features
data1 = (
    data.stack(level=0, future_stack=True)  
      .rename_axis(["Date", "Ticker"])
      .reset_index()
)

data1 = data1[["Date","Ticker","Open","High","Low","Close","Volume"]]
data1 = data1.sort_values(["Ticker","Date"]).reset_index(drop=True)

data1.to_csv("stock_data.csv", index=False)
```
